# Auto Tagging Support Tickets Using LLM

## Objective
The objective of this project is to automatically classify support tickets into categories using Large Language Models (LLMs).

```python
!pip install transformers torch
```

# Import Libraries

```python
from transformers import pipeline
```

# Load Zero-Shot Classification Model

```python
classifier = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli"
)
```

# Classify Support Ticket

```python
ticket = "Internet connection is not working after software update"

labels = [
    "network issue",
    "billing",
    "technical issue",
    "account issue"
]

result = classifier(ticket, labels)

result
```

# Top 3 Predicted Tags

```python
top3 = result["labels"][:3]

print(top3)
```

# Few-Shot Learning Example

```python
examples = """
Ticket: Cannot login to account
Tag: account issue

Ticket: Payment failed
Tag: billing

Ticket: Wifi disconnects repeatedly
Tag: network issue
"""

print(examples)
```

# Conclusion

The Large Language Model successfully classified support tickets into relevant categories using zero-shot classification techniques. Few-shot prompting can further improve prediction accuracy.
