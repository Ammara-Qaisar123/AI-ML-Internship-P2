# Import Required Libraries

In this step, we import the necessary libraries required for document loading, embeddings, and vector database creation.

```python
from langchain_community.document_loaders import TextLoader
from langchain_huggingface.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
```

# Create Knowledge Base

A small text-based knowledge base is created for the chatbot. The chatbot will retrieve answers from this document.

```python
text = """
Artificial Intelligence is the simulation of human intelligence.

Machine Learning is a subset of Artificial Intelligence.

Deep Learning uses neural networks for complex tasks.

Natural Language Processing helps computers understand human language.
"""

with open("knowledge.txt", "w") as f:
    f.write(text)
```

#Load Document

```python
loader = TextLoader("knowledge.txt")

docs = loader.load()

docs
```

# Create Embeddings

Embeddings convert textual information into numerical vectors that help the chatbot understand semantic meaning.

```python
embeddings = HuggingFaceEmbeddings()
```

# Create Vector Database

FAISS vector database is used to store document embeddings for efficient similarity search and retrieval.

```python
db = FAISS.from_documents(docs, embeddings)
```

# Ask Question from Chatbot

The chatbot retrieves the most relevant information from the knowledge base based on the user query.

```python
query = "What is Machine Learning?"

results = db.similarity_search(query)

print(results[0].page_content)
```

# Context Memory

Context memory allows the chatbot to remember previous conversations and maintain conversational flow.

```python
!pip install langchain --force-reinstall

from langchain_community.memory import ConversationBufferMemory

memory = ConversationBufferMemory()

memory.save_context(
    {"input":"What is AI?"},
    {"output":"AI is simulation of human intelligence."}
)

memory.load_memory_variables({})
```

# Conclusion

The RAG-based chatbot successfully retrieved relevant information from the knowledge base using embeddings and vector similarity search. Context memory improved conversational continuity and user interaction.
